```
X = df.drop(['class'], axis=1)

y = df['class']
```

This separates the dataset into features (X) and target (y).

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42)
```

This splits the data into training and testing sets, with 33% used for testing and 67% for training.

```
X_train.shape, X_test.shape
```

This shows the number of rows and columns in the training and testing feature sets.

```
X_train.dtypes
```

This shows the data type of each column in the training feature set.

```
X_train.head()
```

This displays the first 5 rows of the training feature set.

```
!pip install category_encoders
```

```
import category_encoders as ce
```

This imports the Category Encoders library and assigns it the shortcut name ce for encoding categorical variables.

```
encoder = ce.OrdinalEncoder(cols=['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety'])


X_train = encoder.fit_transform(X_train)

X_test = encoder.transform(X_test)
```

This code converts categorical features into numerical values using ordinal encoding for both training and testing data.

```
X_train.head()
```

This displays the first 5 rows of the encoded training dataset.

```
X_test.head()
```

This displays the first 5 rows of the encoded testing dataset.

```
from sklearn.tree import DecisionTreeClassifier
```

This imports the Decision Tree Classifier model from scikit-learn for building a classification model.

```
clf_gini = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=0)
clf_gini.fit(X_train, y_train)
```

This creates a Decision Tree model using the Gini impurity measure, limits the tree depth to 3, and trains it using the training data.

```
y_pred_gini = clf_gini.predict(X_test)
```

This uses the trained model to predict the target values for the test dataset.

```
from sklearn.metrics import accuracy_score

print('Model accuracy score with criterion gini index: {0:0.4f}'. format(accuracy_score(y_test, y_pred_gini)))
```

This calculates and prints the accuracy of the decision tree model by comparing predicted values with actual test values.

```
y_pred_train_gini = clf_gini.predict(X_train)

y_pred_train_gini
```

This uses the trained model to predict the target values for the training dataset.

```
print('Training-set accuracy score: {0:0.4f}'. format(accuracy_score(y_train, y_pred_train_gini)))
```

This calculates and prints the accuracy of the model on the training data.

```
print('Training set score: {:.4f}'.format(clf_gini.score(X_train, y_train)))

print('Test set score: {:.4f}'.format(clf_gini.score(X_test, y_test)))
```

This prints the model’s accuracy on both training and test datasets using the built-in score() method.

```
plt.figure(figsize=(12,8))

from sklearn import tree

tree.plot_tree(clf_gini.fit(X_train, y_train))
```

This code visualizes the trained decision tree model, showing how it splits the data to make predictions.

```
import graphviz
dot_data = tree.export_graphviz(clf_gini, out_file=None,
                              feature_names=X_train.columns,
                              class_names=y_train,
                              filled=True, rounded=True,
                              special_characters=True)

graph = graphviz.Source(dot_data)

graph
```

This code converts the trained decision tree into a Graphviz format and visualizes it as a clear tree diagram showing decision rules and class predictions.

```
clf_en = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=0)
clf_en.fit(X_train, y_train)
```

This creates and trains a Decision Tree model using entropy (information gain) as the splitting criterion.

```
y_pred_en = clf_en.predict(X_test)
```

This uses the trained entropy-based decision tree to predict values for the test data.

```
from sklearn.metrics import accuracy_score

print('Model accuracy score with criterion entropy: {0:0.4f}'. format(accuracy_score(y_test, y_pred_en)))
```

This calculates and prints the accuracy of the entropy-based decision tree model on the test data.

```
y_pred_train_en = clf_en.predict(X_train)

y_pred_train_en
```

This uses the trained entropy-based model to predict outcomes for the training data.

```
print('Training-set accuracy score: {0:0.4f}'. format(accuracy_score(y_train, y_pred_train_en)))
```

This calculates and prints the accuracy of the entropy-based decision tree model on the training data.

```
print('Training set score: {:.4f}'.format(clf_en.score(X_train, y_train)))

print('Test set score: {:.4f}'.format(clf_en.score(X_test, y_test)))
```

This prints the accuracy of the entropy-based decision tree model on both the training set and the test set.

```
plt.figure(figsize=(12,8))

from sklearn import tree

tree.plot_tree(clf_en.fit(X_train, y_train))
```

This code visualizes the entropy-based decision tree model, showing how the data is split at each decision node to make predictions.

```
import graphviz
dot_data = tree.export_graphviz(clf_en, out_file=None,
                              feature_names=X_train.columns,
                              class_names=y_train,
                              filled=True, rounded=True,
                              special_characters=True)

graph = graphviz.Source(dot_data)

graph
```

This code converts the entropy-based decision tree model into a Graphviz format and displays it as a visual tree diagram showing decision rules and class splits.

```
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_en)

print('Confusion matrix\n\n', cm)
```

This code generates and prints the confusion matrix, which shows the number of correct and incorrect predictions made by the model.

```
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_en))
```

This code prints a classification report showing precision, recall, f1-score, and support for each class in the model’s predictions.
